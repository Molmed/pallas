```python
import os

import pandas as pd
import plotly.graph_objects as go

REPORT_OUTPUT_DIR = "/home/mariya/Development/pallas/reports/output/evaluation/bimodal_labeled"

# Create dir if it doesnt exist
if not os.path.exists(REPORT_OUTPUT_DIR):
    os.makedirs(REPORT_OUTPUT_DIR)

colors_tsv = "/home/mariya/Development/pallas/configs/class_colors.tsv"

BLUE="#667ec8"
RED="#d64a51"
YELLOW="#d6ce2f"
GREEN="#90c23a"
LIGHTBLUE="#60acda"
PURPLE="#9454b0"

# Read in pallette
palette_df = pd.read_csv(colors_tsv, sep="\t", header=None, names=["class", "color"])
palette_df
```

```python

df = pd.read_csv("/home/mariya/Development/pallas/reports/output/predictions/bimodal/bimodal_predictions_labeled.csv", index_col=0)
df
```

```python
def sankey(
    df: pd.DataFrame,
    gex_col: str = "gex_pred",
    dualmodal_col: str = "dualmodal_pred",
    dnam_col: str = "dnam_pred",
    title: str = "GEX / True subtype / DNAm predictions",
) -> go.Figure:
    """Create a Sankey with GEX predictions on the left, true subtype in the middle, and DNAm predictions on the right."""
    required_columns = [gex_col, dualmodal_col, dnam_col]
    missing_columns = [column for column in required_columns if column not in df.columns]
    if missing_columns:
        raise ValueError(f"Missing prediction columns: {missing_columns}")

    palette_by_class = {
        str(class_name).strip(): str(color).strip()
        for class_name, color in zip(palette_df["class"], palette_df["color"])
    }
    default_color = "#BDBDBD"

    left_labels = sorted(df[gex_col].dropna().astype(str).unique())
    middle_labels = sorted(df[dualmodal_col].dropna().astype(str).unique())
    right_labels = sorted(df[dnam_col].dropna().astype(str).unique())

    # Keep one node per side so a class can appear independently in each layer.
    node_labels = left_labels + middle_labels + right_labels
    node_x = (
        [0.02] * len(left_labels)
        + [0.50] * len(middle_labels)
        + [0.98] * len(right_labels)
    )
    node_y = (
        [(index + 0.5) / (len(left_labels) + 1.5) for index in range(len(left_labels))]
        + [(index + 0.5) / (len(middle_labels) + 1.5) for index in range(len(middle_labels))]
        + [(index + 0.5) / (len(right_labels) + 1.5) for index in range(len(right_labels))]
    )
    node_colors = [palette_by_class.get(label, default_color) for label in node_labels]

    node_index = {
        ("GEX", label): index for index, label in enumerate(left_labels)
    }
    node_index.update(
        {
            ("True subtype", label): len(left_labels) + index
            for index, label in enumerate(middle_labels)
        }
    )
    node_index.update(
        {
            ("DNAm", label): len(left_labels) + len(middle_labels) + index
            for index, label in enumerate(right_labels)
        }
    )

    left_to_middle = df.groupby([gex_col, dualmodal_col], dropna=True).size()
    middle_to_right = df.groupby([dualmodal_col, dnam_col], dropna=True).size()

    sources, targets, values, link_colors = [], [], [], []

    def rgba(hex_color: str, alpha: float = 0.45) -> str:
        color = hex_color.lstrip("#")
        red, green, blue = (int(color[index:index + 2], 16) for index in (0, 2, 4))
        return f"rgba({red},{green},{blue},{alpha})"

    for (gex, true_subtype), count in left_to_middle.items():
        gex = str(gex)
        true_subtype = str(true_subtype)
        sources.append(node_index[("GEX", gex)])
        targets.append(node_index[("True subtype", true_subtype)])
        values.append(int(count))
        link_colors.append(rgba(palette_by_class.get(gex, default_color)))

    for (true_subtype, dnam), count in middle_to_right.items():
        true_subtype = str(true_subtype)
        dnam = str(dnam)
        sources.append(node_index[("True subtype", true_subtype)])
        targets.append(node_index[("DNAm", dnam)])
        values.append(int(count))
        link_colors.append(rgba(palette_by_class.get(true_subtype, default_color)))

    fig = go.Figure(
        go.Sankey(
            arrangement="snap",
            node=dict(
                pad=18,
                thickness=18,
                line=dict(color="black", width=0.5),
                label=node_labels,
                color=node_colors,
                x=node_x,
                y=node_y,
            ),
            link=dict(
                source=sources,
                target=targets,
                value=values,
                color=link_colors,
            ),
        )
    )
    fig.update_layout(
        title_text=title,
        title_font=dict(size=20),
        height=700 + 45 * max(len(left_labels), len(middle_labels), len(right_labels)),
        margin=dict(l=20, r=20, t=70, b=80),
        font=dict(family="Arial, sans-serif", size=16, color="black"),
    )
    return fig


fig = sankey(
    df,
    gex_col="GEX sPLS-DA prediction",
    dualmodal_col="True subtype",
    dnam_col="DNAm sPLS-DA prediction",
    title="GEX / True subtype / DNAm predictions",
)
fig_path = os.path.join(REPORT_OUTPUT_DIR, "two_way_sankey.svg")
try:
    fig.write_image(fig_path)
except Exception:
    html_path = os.path.splitext(fig_path)[0] + ".html"
    fig.write_html(html_path)
    print(f"SVG export unavailable; saved interactive Sankey to {html_path}")
fig.show()
```

```python
# Add an upset plot
import matplotlib.pyplot as plt
from upsetplot import UpSet, from_indicators

true_subtype_col = "True subtype"
concordance = pd.DataFrame(
    {
        "GEX": df["GEX sPLS-DA prediction"].eq(df[true_subtype_col]),
        "DNAm": df["DNAm sPLS-DA prediction"].eq(df[true_subtype_col]),
        "Dual-modal": df["DIABLO sPLS-DA prediction"].eq(df[true_subtype_col]),
    },
    index=df.index,
)

upset_data = from_indicators(list(concordance.columns), data=concordance)
upset = UpSet(
    upset_data,
    subset_size="count",
    sort_by="cardinality",
    show_counts=True,
    facecolor=BLUE,
)
upset.plot()
plt.suptitle("Concordance with true subtype", y=1.02, fontsize=16)
plt.tight_layout()
upset_path = os.path.join(REPORT_OUTPUT_DIR, "upset_concordance.png")
plt.savefig(upset_path, dpi=300, bbox_inches="tight")
plt.show()
print(f"Saved UpSet plot to {upset_path}")
```

```python
# Sankey: classifier-to-classifier concordance
classifier_df = df.rename(
    columns={
        "GEX sPLS-DA prediction": "GEX",
        "DNAm sPLS-DA prediction": "DNAm",
        "DIABLO sPLS-DA prediction": "Dual-modal",
    }
)

sankey_classifier_df = classifier_df[["GEX", "DNAm", "Dual-modal"]].dropna()

fig = sankey(
    sankey_classifier_df,
    gex_col="GEX",
    dualmodal_col="Dual-modal",
    dnam_col="DNAm",
    title="GEX / Dual-modal / DNAm classifier concordance",
)

classifier_sankey_path = os.path.join(
    REPORT_OUTPUT_DIR, "classifier_concordance_sankey.svg"
)
try:
    fig.write_image(classifier_sankey_path)
except Exception:
    classifier_sankey_path = os.path.splitext(classifier_sankey_path)[0] + ".html"
    fig.write_html(classifier_sankey_path)

fig.show()
```

```python
# UpSet: classifier concordance based on majority calls
from upsetplot import UpSet, from_memberships

classifier_df = df.rename(
    columns={
        "GEX sPLS-DA prediction": "GEX",
        "DNAm sPLS-DA prediction": "DNAm",
        "DIABLO sPLS-DA prediction": "Dual-modal",
    }
).copy()

classifiers = ["GEX", "DNAm", "Dual-modal"]
classifier_df = classifier_df[classifiers].dropna()

# A classifier belongs to a sample's set when it matches that sample's
# majority prediction.
classifier_df["majority"] = classifier_df[classifiers].mode(axis=1)[0]

memberships = []
for _, row in classifier_df.iterrows():
    memberships.append(
        [classifier for classifier in classifiers if row[classifier] == row["majority"]]
    )

upset_data = from_memberships(memberships)

fig = plt.figure(figsize=(10, 6))
upset = UpSet(
    upset_data,
    subset_size="count",
    show_counts=True,
    sort_by="cardinality",
    sort_categories_by=None,
    facecolor=BLUE,
)
upset.plot(fig)

plt.suptitle(
    "Classifier concordance (GEX / DNAm / Dual-modal)",
    y=1.02,
)

class_upset_path = os.path.join(
    REPORT_OUTPUT_DIR,
    "classifier_concordance_upset.svg",
)
plt.savefig(class_upset_path, dpi=150, bbox_inches="tight")
plt.show()
print(f"Saved UpSet plot to {class_upset_path}")
```
