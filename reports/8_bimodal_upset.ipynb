```python
import pandas as pd
from upsetplot import UpSet, from_memberships
import matplotlib.pyplot as plt
```

```python
colors_tsv = "/home/mariya/Development/pallas/configs/class_colors.tsv"

# Read in pallette
palette_df = pd.read_csv(colors_tsv, sep="\t", header=None, names=["class", "color"])
palette_df
```

```python

df = pd.read_csv("/home/mariya/Development/pallas/reports/output/predictions/bimodal/unlabeled_predictions.csv", index_col=0)
df
```

```python
# Grab the first three cols and rename them to "GEX", "DNAm", and "DIABLO"
df = df.iloc[:, 1:4]
df.columns = ["GEX", "DNAm", "DIABLO"]
df
```

```python

classifiers = ["GEX", "DNAm", "DIABLO"]

# Build per-sample memberships: a sample belongs to a set if all
# classifiers in that set agree with the majority call for that sample.
# Here we define "concordant for classifier X" as: X matches the majority vote.
df["majority"] = df[classifiers].mode(axis=1)[0]

memberships = []
for _, row in df.iterrows():
    member_of = [c for c in classifiers if row[c] == row["majority"]]
    memberships.append(member_of)

upset_data = from_memberships(memberships)

fig = plt.figure(figsize=(10, 6))
upset = UpSet(
    upset_data,
    subset_size="count",
    show_counts=True,
    sort_by="cardinality",      # puts "all three agree" first
    sort_categories_by=None,
)
upset.plot(fig)

plt.suptitle("Classifier concordance (GEX / DNAm / DIABLO)", y=1.02)
plt.savefig("upset_concordance.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python

# Collect all unique predicted classes
all_classes = pd.unique(df[classifiers].values.ravel())

# For each sample, which classes were predicted (by any classifier)?
memberships = []
for _, row in df.iterrows():
    predicted = list(set(row[c] for c in classifiers))
    memberships.append(sorted(predicted))  # sort for consistent set hashing

upset_data = from_memberships(memberships, data=df.index)

fig = plt.figure(figsize=(12, 6))
upset = UpSet(
    upset_data,
    subset_size="count",
    show_counts=True,
    sort_by="cardinality",
    sort_categories_by="-input",   # preserves your class order
)
upset.plot(fig)

plt.suptitle("Subtype class memberships across classifiers", y=1.02)
plt.savefig("upset_class_memberships.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Expected columns:
# df: columns GEX, DNAm, DIABLO
# palette_df: columns subtype, color

classifiers = ["GEX", "DNAm", "DIABLO"]

# Build a subtype -> color mapping from palette_df
color_map = dict(zip(palette_df["class"], palette_df["color"]))

fig, axes = plt.subplots(1, 3, figsize=(18, 6), constrained_layout=True)

for ax, classifier in zip(axes, classifiers):
    counts = df[classifier].value_counts()

    labels = counts.index.tolist()
    colors = [color_map.get(subtype, "#cccccc") for subtype in labels]

    ax.pie(
        counts.values,
        labels=labels,
        colors=colors,
        autopct=lambda pct: f"{pct:.1f}%" if pct >= 3 else "",
        startangle=90,
        counterclock=False,
        rotatelabels=True,
        labeldistance=1.12,
        wedgeprops={"edgecolor": "white", "linewidth": 1},
        textprops={"fontsize": 9},
    )
    ax.set_title(f"{classifier} predictions", fontsize=12)
    ax.axis("equal")

plt.suptitle("Subtype breakdown of predictions by classifier", y=1.02, fontsize=14)
plt.show()
```
