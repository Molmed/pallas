```python
import numpy as np
import pandas as pd
import os
import dotenv
```

```python
dotenv.load_dotenv()

DATASETS_PATH = os.getenv("DATASETS_PATH")
MODELS_DIR = os.getenv("MODELS_DIR")
OUTPUT_DIR = os.path.join(os.getenv("REPORTS_DIR"), "cross_val_metrics")

# Create output directory if it doesn't exist
os.makedirs(OUTPUT_DIR, exist_ok=True)

from helpers import load_model, load_dataset, load_merged_test_datasets, classifier_level_metrics, cross_val_metrics
```

```python
def cross_val_modality(modality):
    model = load_model(modality)
    X_train, y_train = load_dataset(modality, "train")
    macro_cv_metric, class_cv_metrics = cross_val_metrics(model, X_train, y_train, model.classes_.tolist())
    return macro_cv_metric, class_cv_metrics
```
