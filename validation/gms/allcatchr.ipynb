```python
import pandas
```

```python
GEX_FILE="/home/mariya/Data/pallas/raw/validation/ClinicalGenomicsUppsala/GeneCounts.GMS-Barn.table.csv"
OUTPUT_FILE="/home/mariya/Data/pallas/gms_predictions/allcatchr/gms.csv"
gex = pandas.read_csv(GEX_FILE, index_col=0)
gex.head()
```

```python
GEX_FILE_2="/home/mariya/Data/pallas/raw/validation/ClinicalGenomicsUppsala/GeneCounts2.out"
gex2 = pandas.read_csv(GEX_FILE_2, index_col=0)
gex2.head()
```

```python
# Append gex2 to gex
gex = pandas.concat([gex, gex2], axis=1)
gex
```

```python
# Name index to gene
gex.index.name = "gene"

# Check for NA values
print(gex.isnull().sum())
```

```python
# What rows have NA values?
na_rows = gex.index[gex.isnull().any(axis=1)]
print("Rows with NA values:", na_rows)
```

```python
# Remove them
gex = gex.dropna()
gex
```

```python


# Save to output file
gex.to_csv(OUTPUT_FILE)
```
