```python
import pandas
from gene_thesaurus import GeneThesaurus
```

```python
GEX_FILE="/home/mariya/Data/pallas/raw/validation/ClinicalGenomicsUppsala/GeneCounts.GMS-Barn.table.csv"
OUTPUT_FILE="/home/mariya/Data/pallas/gms_predictions/allsorts/gms.csv"
gex = pandas.read_csv(GEX_FILE, index_col=0)
gex.head()
```

```python
GEX_FILE_2="/home/mariya/Data/pallas/raw/validation/ClinicalGenomicsUppsala/GeneCounts2.out"
gex2 = pandas.read_csv(GEX_FILE_2, index_col=0)
gex2.head()
```

```python
# Append gex2 to gex
gex = pandas.concat([gex, gex2], axis=1)
gex
```

```python
gt = GeneThesaurus()

# Translate Ensembl IDs to gene symbols
translated = gt.translate_genes(gex.index.tolist(), source="ensembl_id", target="symbol")
translated
```

```python
# Create new df with translated gene symbols as index
# Use old values if any are missing
gex_translated = gex.copy()
gex_translated.index = [translated.get(gene, gene) for gene in gex.index]
gex_translated = gex_translated.T
gex_translated.head()
```

```python
# Save to output file
gex_translated.to_csv(OUTPUT_FILE)
```
