```python
import pandas as pd

VAL_NONE = "'--"

followup_path = '/home/mariya/Data/gdc/all/all_gex/clinical.cart.2025-10-08/follow_up.tsv'
samplesheet_path = '/home/mariya/Data/gdc/all/all_gex/gdc_sample_sheet.2025-10-08.tsv'
output_file_path = '/home/mariya/Data/gdc/all/all_gex/gdc.all.gex.subtyped.tsv'
manifest_file_path = '/home/mariya/Data/gdc/all/all_gex/gdc_manifest.2025-10-08.102143.txt'
followup_df = pd.read_csv(followup_path, sep='\t')
followup_df.head(), followup_df.shape
```

```python
# Get all columns that contain values other than VAL_NONE
non_empty_cols = followup_df.columns[followup_df.ne(VAL_NONE).any()]
non_empty_cols, non_empty_cols.size
```

```python
# Get only data points where 'molecular_tests.test_result' is 'Positive'
positive_tests_df = followup_df[followup_df['molecular_tests.test_result'] == 'Positive']
positive_tests_df, positive_tests_df.shape
```

```python
# Get new non empty cols
non_empty_cols = positive_tests_df.columns[positive_tests_df.ne(VAL_NONE).any()]
non_empty_cols, non_empty_cols.size
```

```python
# Get only non empty columns
refined_df = positive_tests_df[non_empty_cols]
refined_df, refined_df.shape
```

```python
# Get unique cases.case_id
unique_case_ids = refined_df['cases.case_id'].unique()
unique_case_ids, unique_case_ids.size
```

```python
# See all cases where molecular_tests.variant_type is 'Fusion'
fusion_cases_df = refined_df[refined_df['molecular_tests.variant_type'] == 'Fusion']
fusion_cases_df, fusion_cases_df.shape
```

```python
# Get unique values for 'molecular_tests.gene_symbol' and 'molecular_tests.second_gene_symbol'
fusion_cases_df['molecular_tests.gene_symbol'].unique(), fusion_cases_df['molecular_tests.second_gene_symbol'].unique()
```

```python
# See all cases where molecular_tests.variant_type is 'Rearrangement'
rearrangement_cases_df = refined_df[refined_df['molecular_tests.variant_type'] == 'Rearrangement']
rearrangement_cases_df, rearrangement_cases_df.shape
```

```python
# Get unique values for 'molecular_tests.gene_symbol' and 'molecular_tests.second_gene_symbol'
rearrangement_cases_df['molecular_tests.gene_symbol'].unique(), rearrangement_cases_df['molecular_tests.second_gene_symbol'].unique()
```

```python
# See all cases where molecular_tests.variant_type is 'Translocation'
translocation_cases_df = refined_df[refined_df['molecular_tests.variant_type'] == 'Translocation']
translocation_cases_df, translocation_cases_df.shape
```

```python
# Get unique values for 'molecular_tests.gene_symbol' and 'molecular_tests.second_gene_symbol'
translocation_cases_df['molecular_tests.gene_symbol'].unique(), translocation_cases_df['molecular_tests.second_gene_symbol'].unique()
```

```python
# Find all rows where 'molecular_tests.gene_symbol' is 'ETV6'
# Then, of those cases.submitter_id, how many have a row where 'molecular_tests.gene_symbol' is 'RUNX1'?
etv6_cases_df = refined_df[refined_df['molecular_tests.gene_symbol'] == 'ETV6']
etv6_case_ids = etv6_cases_df['cases.case_id'].unique()
runx1_cases_df = refined_df[(refined_df['molecular_tests.gene_symbol'] == 'RUNX1') & (refined_df['cases.case_id'].isin(etv6_case_ids))]
runx1_cases_df, runx1_cases_df.shape
# for all these, set new subtype col to 'ETV6::RUNX1' in refined_df
refined_df['subtype'] = None
refined_df.loc[refined_df['cases.case_id'].isin(runx1_cases_df['cases.case_id']), 'subtype'] = 'ETV6::RUNX1'
refined_df, refined_df.shape
```

```python
def determine_subtype(row):
    gene1 = row['molecular_tests.gene_symbol']
    gene2 = row['molecular_tests.second_gene_symbol']
    variant_type = row['molecular_tests.variant_type']

    if variant_type == 'Translocation':
        if (gene1 == 'TCF3' and gene2 == 'PBX1'):
            return 'TCF3::PBX1'

    if variant_type == 'Rearrangement':
        if gene1 == 'KMT2A':
            return 'KMT2A-r'

    if variant_type == 'Fusion':
        if (gene1 == 'ETV6' and gene2 == 'RUNX1'):
            return 'ETV6::RUNX1'
        elif (gene1 == 'BCR' and gene2 == 'ABL1'):
            return 'BCR::ABL1'

refined_df['subtype'] = refined_df.apply(determine_subtype, axis=1)
```

```python
refined_df['subtype'].value_counts(), refined_df.shape
```

```python
# Create a new df with columns: project.project_id	cases.case_id	cases.submitter_id subtype
# Create only one entry per case_id with subtype if available
refined_df = refined_df[['project.project_id', 'cases.submitter_id', 'subtype']]
refined_df, refined_df.shape
```

```python
# Drop all where subtype is NaN
refined_df = refined_df.dropna(subset=['subtype'])
refined_df, refined_df.shape
```

```python
refined_df = refined_df.drop_duplicates()
refined_df, refined_df.shape
```

```python
# Show all rows where cases.submitter_id is duplicated
dupes = refined_df[refined_df.duplicated(subset=['cases.submitter_id'], keep=False)].sort_values(by='cases.submitter_id')
dupes, dupes.shape
```

```python
# Get samplesheet
samplesheet_df = pd.read_csv(samplesheet_path, sep='\t')
samplesheet_df.head(), samplesheet_df.shape
```

```python
cols_to_keep = ['Case ID', 'Sample ID', 'File Name', 'Tissue Type', 'Tumor Descriptor', 'Specimen Type']
samplesheet_df = samplesheet_df[cols_to_keep]
samplesheet_df
```

```python
# Join cases.submitter_id in merged_df with Case ID in samplesheet_df to get file names
merged_df = refined_df.merge(samplesheet_df[cols_to_keep], left_on='cases.submitter_id', right_on='Case ID', how='left')
merged_df.shape
merged_df
```

```python
merged_df = merged_df.drop(columns=['cases.submitter_id'])
merged_df = merged_df.rename(columns={'project.project_id': 'Project Id'})
merged_df
```

```python
# Show rows where Case ID, Tissue Type and Tumor Descriptor are all duplicated
dupes = merged_df[merged_df.duplicated(subset=['Case ID', 'Tissue Type', 'Tumor Descriptor'], keep=False)].sort_values(by=['Case ID', 'Tissue Type', 'Tumor Descriptor'])
dupes
```

```python
# Create a new column "Replicate" that indicates whether the row is a replicate
# but do not mark the first of each set as true
merged_df['Replicate'] = merged_df.duplicated(subset=['Case ID', 'Tissue Type', 'Tumor Descriptor'], keep='first')
merged_df
```

```python
# Create new column Unique Sample ID which contains Sample ID + _R1, _R2 etc for replicates
# First, create a helper column that numbers the replicates
merged_df['Replicate Number'] = merged_df.groupby(['Case ID', 'Tissue Type', 'Tumor Descriptor']).cumcount()
merged_df['Unique Sample ID'] = merged_df.apply(lambda row: f"{row['Sample ID']}_R{row['Replicate Number']}" if row['Replicate'] else row['Sample ID'], axis=1)
merged_df
```

```python
# See if there are any duplicated Unique Sample ID
merged_df[merged_df.duplicated(subset=['Unique Sample ID'], keep=False)], merged_df.shape
```

```python
merged_df.to_csv(output_file_path, sep='\t', index=False)
```

```python
# Read in manifest
manifest_df = pd.read_csv(manifest_file_path, sep='\t')
manifest_df.head(), manifest_df.shape
```

```python
# Filter out rows in manifest where filename is not in merged_df['File Name']
manifest_df = manifest_df[manifest_df['filename'].isin(merged_df['File Name'])]
manifest_df, manifest_df.shape
```

```python
# Save filtered manifest
manifest_df.to_csv(manifest_file_path.replace('.txt', '.filtered.txt'), sep='\t', index=False)
```
