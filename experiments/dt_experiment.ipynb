```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
import pandas as pd
import os
import joblib
from dotenv import load_dotenv

load_dotenv()

X_train_path = os.getenv("X_TRAIN_CSV_PATH")
y_train_path = os.getenv("Y_TRAIN_TSV_PATH")
X_test_path = os.getenv("X_TEST_CSV_PATH")
y_test_path = os.getenv("Y_TEST_TSV_PATH")
output_dir = os.getenv("OUTPUT_DIR")
random_seed = int(os.getenv("RANDOM_SEED"))
n_jobs = int(os.getenv("N_JOBS", -1))
selected_features_path = os.getenv("SELECTED_FEATURES_PATH")
rf_model_path = os.getenv('RF_MODEL_PATH')
N_TOP_FEATURES = int(os.getenv("N_TOP_FEATURES", 500))


# Print configuration
print(f"X_train_path: {X_train_path}")
print(f"y_train_path: {y_train_path}")
print(f"X_test_path: {X_test_path}")
print(f"y_test_path: {y_test_path}")
print(f"output_dir: {output_dir}")
print(f"random_seed: {random_seed}")
print(f"n_jobs: {n_jobs}")
print(f"selected_features_path: {selected_features_path}")
print(f"rf_model_path: {rf_model_path}")
print(f"N_TOP_FEATURES: {N_TOP_FEATURES}")
```

```python
# Read selected features
selected_features = joblib.load(selected_features_path)

# Read rf model
rf_model = joblib.load(rf_model_path)

# Join selected features and feature importances
feature_importances = pd.DataFrame({
    "feature": selected_features,
    "importance": rf_model.feature_importances_
})
feature_importances_ranked = feature_importances.sort_values(by="importance", ascending=False)
feature_importances_ranked
```

```python
# Read in data
X_train = pd.read_csv(X_train_path, index_col=0)
y_train = pd.read_csv(y_train_path, sep="\t", index_col=0)
X_test = pd.read_csv(X_test_path, index_col=0)
y_test = pd.read_csv(y_test_path, sep="\t", index_col=0)
```

```python

# align the indices
X_train = X_train.loc[y_train.index]
X_train

X_test = X_test.loc[y_test.index]
X_test
```

```python
# Get the top selected genes from feature_importances_ranked
top_genes = feature_importances_ranked['feature'].head(N_TOP_FEATURES).tolist()

# Subset the training and test sets
X_train_selected = X_train[top_genes]
X_test_selected = X_test[top_genes]

# Initialize decision tree
dt = DecisionTreeClassifier(
    criterion="gini",     # or "entropy"
    max_depth=None,       # can tune this to prevent overfitting
    min_samples_split=2,  # minimum samples required to split
    min_samples_leaf=1,   # minimum samples at a leaf
    random_state=random_seed,
    class_weight='balanced'  # handle class imbalance
)

# Train
dt.fit(X_train_selected, y_train)

# Evaluate
y_pred = dt.predict(X_test_selected)
print(classification_report(y_test, y_pred))
```
