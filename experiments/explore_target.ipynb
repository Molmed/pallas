```python
import pandas as pd

VAL_NONE = "'--"

followup_path = '/home/mariya/Data/gdc/aml/aml_gex/follow_up.tsv'
output_file_path = '/home/mariya/Data/gdc/aml/aml_gex/follow_up_refined.tsv'
followup_df = pd.read_csv(followup_path, sep='\t')
followup_df.head(), followup_df.shape
```

```python
# Get all columns that contain values other than VAL_NONE
non_empty_cols = followup_df.columns[followup_df.ne(VAL_NONE).any()]
non_empty_cols, non_empty_cols.size
```

```python
# Get only data points where 'molecular_tests.test_result' is 'Positive'
positive_tests_df = followup_df[followup_df['molecular_tests.test_result'] == 'Positive']
positive_tests_df, positive_tests_df.shape
```

```python
# Get new non empty cols
non_empty_cols = positive_tests_df.columns[positive_tests_df.ne(VAL_NONE).any()]
non_empty_cols, non_empty_cols.size
```

```python
# Get only non empty columns
refined_df = positive_tests_df[non_empty_cols]
refined_df, refined_df.shape
```

```python
# Get unique cases.case_id
unique_case_ids = refined_df['cases.case_id'].unique()
unique_case_ids, unique_case_ids.size
```

```python
# See all cases where molecular_tests.variant_type is 'Fusion'
fusion_cases_df = refined_df[refined_df['molecular_tests.variant_type'] == 'Fusion']
fusion_cases_df, fusion_cases_df.shape
```

```python
# Get unique values for 'molecular_tests.gene_symbol' and 'molecular_tests.second_gene_symbol'
fusion_cases_df['molecular_tests.gene_symbol'].unique(), fusion_cases_df['molecular_tests.second_gene_symbol'].unique()
```

```python
# Create a new subtype column based on gene_symbol and second_gene_symbol
def determine_subtype(row):
    gene1 = row['molecular_tests.gene_symbol']
    gene2 = row['molecular_tests.second_gene_symbol']

    if (gene1 == 'PML' and gene2 == 'RARA'):
        return 'APL-PML::RARA'
    elif (gene1 == 'RUNX1' and gene2 == 'RUNX1T1'):
        return 'RUNX1::RUNX1T1'
    elif (gene1 == 'CBFB' and gene2 == 'MYH11'):
        return 'CBFB::MYH11'
    elif (gene1 == 'BCR' and gene2 == 'ABL1'):
        return 'BCR::ABL1 AML'
    elif (gene1 == 'KMT2A' or gene2 == 'KMT2A'):
        return 'KMT2A-r'
    elif (gene1 == 'DEK' and gene2 == 'NUP214'):
        return 'DEK::NUP214'
    elif(gene1 == 'NUP98' or gene2 == 'NUP98'):
        return 'NUP98-r'
    elif(gene1 == 'MECOM' or gene2 == 'MECOM'):
        return 'MECOM-r'
    elif(gene1 == 'KAT6A' or gene2 == 'KAT6A'):
        return 'KAT6A-r'
    elif(gene1 == 'PICALM' and gene2 == 'MLLT10'):
        return 'PICALM::MLLT10'
    elif(gene1 == 'CBFA2T3' and gene2 == 'GLIS2'):
        return 'CBFA2T3::GLIS2'
    elif(gene1 == 'FUS' and gene2 == 'ERG'):
        return 'FUS::ERG'
    elif(gene1 == 'ETV6' or gene2 == 'ETV6'):
        return 'ETV6-r'
    elif(gene1 == 'RBM15' and gene2 == 'MKL1'):
        return 'RBM15::MKL1'
refined_df['subtype'] = refined_df.apply(determine_subtype, axis=1)
```

```python
# Look at unique values of 'molecular_tests.gene_symbol' and 'molecular_tests.second_gene_symbol'
# for items where subtype is empty and molecular_tests.variant_type is 'Mutation, NOS'
refined_df[refined_df['subtype'].isna()][['molecular_tests.gene_symbol', 'molecular_tests.second_gene_symbol']].drop_duplicates()
```

```python
# Mark items with 'molecular_tests.variant_type' as 'Mutation, NOS' and 'NPM1' in either 'molecular_tests.gene_symbol' or 'molecular_tests.second_gene_symbol' as 'NPM1-mutated'
def determine_mutated(row):
    gene1 = row['molecular_tests.gene_symbol']
    gene2 = row['molecular_tests.second_gene_symbol']
    variant_type = row['molecular_tests.variant_type']

    if (variant_type == 'Mutation, NOS' and (gene1 == 'NPM1' or gene2 == 'NPM1')):
        return 'NPM1alt'
    # Same with CEBPA
    elif (variant_type == 'Mutation, NOS' and (gene1 == 'CEBPA' or gene2 == 'CEBPA')):
        return 'CEBPAalt'
    else:
        return row['subtype']

refined_df['subtype'] = refined_df.apply(determine_mutated, axis=1)
```

```python
# Get unique molecular_tests.chromosomal_translocation values for items where subtype is empty and molecular_tests.variant_type is 'Translocation'
refined_df[refined_df['subtype'].isna() & (refined_df['molecular_tests.variant_type'] == 'Translocation')]['molecular_tests.chromosomal_translocation'].unique()
```

```python
# Mark items with 'molecular_tests.variant_type' as 'Translocation' and known 'molecular_tests.chromosomal_translocation' as their respective subtypes
def determine_translocation_subtype(row):
    if not row['molecular_tests.variant_type'] == 'Translocation':
        return row['subtype']
    if row['molecular_tests.chromosomal_translocation'] == 't(9;22)':
        return 'BCR::ABL1 AML'
    elif row['molecular_tests.chromosomal_translocation'] == 't(9;11)(p22;q23)' or \
        row['molecular_tests.chromosomal_translocation'] == 't(10;11)(p11.2;q23)' or \
        row['molecular_tests.chromosomal_translocation'] == 't(11;19)(q23;p13.1)' or \
        row['molecular_tests.chromosomal_translocation'] == 't(6;11)(q27;q23)' or \
        row['molecular_tests.chromosomal_translocation'] == 't(9;11)':
        return 'KMT2A-r'
    elif row['molecular_tests.chromosomal_translocation'] == 't(8;21)':
        return 'RUNX1::RUNX1T1'
    elif row['molecular_tests.chromosomal_translocation'] == 't(3;5)(q25;q34)':
        return 'NPM1alt'
    elif row['molecular_tests.chromosomal_translocation'] == 't(6;9)':
        return 'DEK::NUP214'
    elif row['molecular_tests.chromosomal_translocation'] == 't(15;17)':
        return 'APL-PML::RARA'
    return row['subtype']

refined_df['subtype'] = refined_df.apply(determine_translocation_subtype, axis=1)
```

```python
# Create a new df with columns: project.project_id	cases.case_id	cases.submitter_id subtype
# Create only one entry per case_id with subtype if available
final_df = refined_df[['project.project_id', 'cases.case_id', 'cases.submitter_id', 'subtype']]
final_df
```

```python
# Get records where subtype is not empty
final_df = final_df[final_df['subtype'].notna()]

# Remove duplicate rows
final_df = final_df.drop_duplicates()
```
